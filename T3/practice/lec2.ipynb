```python
from bs4 import BeautifulSoup
soup=BeautifulSoup(open('politifact.html','r',encoding='UTF-8'))
print(soup.prettify())
```

```python
box=soup.find_all("li",class_="o-listicle__item")
print(len(box))
```

```python
quote=[]
source=[]
date=[]
link=[]
for i in box:
    j=i.find('div',class_="m-statement__quote")
    quote.append(j.find('a').text.strip())
    k=i.find('div',class_="m-statement__meta")
    source.append(k.find('a')['title'])
    date.append(i.find('footer',class_="m-statement__footer").text.split("•")[-1])
    link.append("https://www.politifact.com"+j.find('a')["href"])
for i in quote:
    print(i)
for i in source:
    print(i)
for i in date:
    print(i)
for i in link:
    print(i)
```

```python
import pandas as pd
data=pd.DataFrame({'quote':quote,'source':source,'date':date,'link':link})
data
```

# top 25 imdb

```python
from bs4 import BeautifulSoup
soup=BeautifulSoup(open('IMDb Top 250 Movies.html','r',encoding='UTF-8'))
print(soup.prettify())
```

```python
#movie name
#year
#rating
# <div class="ipc-metadata-list-summary-item__c">
# ipc-title__text
# <span aria-label="IMDb rating: 8.6" class="ipc-rating-star ipc-rating-star--base ipc-rating-star--imdb ratingGroup--imdb-rating" data-testid="ratingGroup--imdb-rating">
box=soup.find_all("div",class_="ipc-metadata-list-summary-item__c")
len(box)
```

```python
movie=[]
year=[]
rating=[]
for i in box:
    year.append(i.find('span',class_="sc-b189961a-8 kLaxqf cli-title-metadata-item").text)
    movie.append(i.find('h3').text.strip())
    rating.append(i.find('span',class_="ipc-rating-star ipc-rating-star--base ipc-rating-star--imdb ratingGroup--imdb-rating").text.split()[0])
for i in year:
    print(i)
for i in movie:
    print(i)
for i in rating:
    print(i)
```

```python
data=pd.DataFrame({'movieName':movie,'Year':year,'rating':rating})
data
```

```python
from bs4 import BeautifulSoup
soup1=BeautifulSoup(open('top250.html','r',encoding='UTF-8'))
print(soup1.prettify())
```

```python
box1=soup1.find_all("td",class_="titleColumn")
len(box1)
```

```python
movie=[]
rating_html=soup1.find_all("td",class_="ratingColumn imdbRating")
rating=[]
for i in rating_html:
    rating.append(i.find('strong').text)
year=[]
for i in box1:
    year.append(i.find('span',class_="secondaryInfo").text)
    movie.append(i.find('a').text)
for i in year:
    print(i)
for i in movie:
    print(i)
for i in rating:
    print(i)
```
